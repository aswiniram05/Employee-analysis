```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
print("data set loaded succesfully")
print("dataset shape:",df.shape)
df.head()
```

```python
salary_analysis = df.groupby('Age')['MonthlyIncome'].mean()

print("Average salary details by Age:")
print(salary_analysis)

print("\nAge with the highest average monthly income:")
print(salary_analysis.idxmax())
```

```python
salary_analysis = df.groupby('Age')['MonthlyIncome'].mean()

print("avg salary details by Age:")
print(salary_analysis)

print("\nhighest avg Age:")
print(salary_analysis.idxmax())
```

```python
attrition_analysis = df.groupby('Department').agg(
    total_Employees=('EmployeeNumber', 'count'),
    Employees_Left=('Attrition', lambda x: (x.str.lower() == 'yes').sum())
)
attrition_analysis['Attrition_Rate'] = (attrition_analysis['Employees_Left'] / attrition_analysis['total_Employees']) * 100
print(attrition_analysis)
```

```python
from numpy._core.numeric import sort
top5 = df.sort_values(by='PerformanceRating', ascending=False).head(5)
print("top5 Employees with highest performance rating:")
display(top5[['EmployeeNumber', 'Department', 'JobRole']])
```

```python
experience_salary=df.groupby('TotalWorkingYears')['MonthlyIncome'].mean()
print("avg salary vs experience salary")
print(experience_salary)
correlation=df[['TotalWorkingYears','MonthlyIncome']].corr()
print("correlation between experience and salary")
print(correlation)
```
